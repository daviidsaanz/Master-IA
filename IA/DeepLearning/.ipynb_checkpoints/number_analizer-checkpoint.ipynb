## Classes 📦

Abstract Layer class defining the common interface for all layers (init, forward, backward, update)

```python
from abc import abstractmethod, ABC
```

```python
class Layer(ABC):
    @abstractmethod
    def init(self): pass

    @abstractmethod
    def forward(self, x): pass

    @abstractmethod
    def backward(self, grad_output): pass

    @abstractmethod
    def update(self, lr): pass
```

Deep class implementing a dense layer: initializes weights W with shape (input_size, output_size) and biases B with shape (1, output_size); W has this shape because each neuron in the layer needs a weight for every input feature, and B has shape (1, output_size) because each neuron has a single bias applied to all examples in the batch. input_size is the number of features per sample (e.g., 784 for MNIST), output_size is the number of neurons in the layer, and the forward computes X @ W + B for a batch of inputs X with shape (batch_size, input_size).

```python
class Deep(Layer):
    def __init__(self, input_size, output_size):
        self.input_size = input_size
        self.output_size = output_size
        self.init()

    def init(self):
        self.W = 2 * np.random.rand(self.input_size, self.output_size) - 1
        # Bias entre 0 y 1
        self.B = np.random.rand(1, self.output_size)

    def forward(self, x):
        self.x = x
        return x @ self.W + self.B

    def backward(self, grad_output):
        grad_input = grad_output @ self.W.T
        self.dW = self.x.T @ grad_output
        self.dB = np.sum(grad_output, axis=0, keepdims=True)
        return grad_input

    def update(self, lr):
        self.W -= self.dW * lr
        self.B -= self.dB * lr
```

x: (batch_size, input_size)
W: (input_size, output_size)
B: (1, output_size)
grad_output: (batch_size, output_size)
grad_input: (batch_size, input_size)
dW: (input_size, output_size)
dB: (1, output_size)



func -> relu, func_deriv -> relu deriv
gradiente total = gradiente que llega (grad_ouput) * derivada de la activacion

```python
class Activation(Layer):
    def __init__(self, func, func_deriv):
        self.func = func
        self.func_deriv = func_deriv
        self.x = None

    def init(self):
        pass

    def forward(self, x):
        self.x = x
        return self.func(x)

    def backward(self, grad_output):
        return grad_output * self.func_deriv(self.x)

    def update(self, lr):
        pass
```

```python
def relu_deriv(x):
    dx = (x > 0).astype(float)
    dx[~np.isfinite(x)] = 0
    return dx


def relu_deriv(x):
    return (x > 0).astype(float)
```

```python
class Model:
    def __init__(self, layers):
        self.sequential = layers

    def forward(self, x):
        for layer in self.sequential:
            x = layer.forward(x)
        return x

    def backward(self, grad_output):
        for layer in reversed(self.sequential):
            grad_output = layer.backward(grad_output)
        return grad_output

    def update(self, lr):
        for layer in self.sequential:
            layer.update(lr)
```

## Implementation 🛠️

```python
def train_step(model, x, y, lr):
    y_pred = model.forward(x)
    error = 2 * (y_pred - y)
    model.backward(error)
    model.update(lr)
    acc = accuracy(y_pred, y)
    print("Accuracy:", acc)

def train_loop(model, data, lr):
    for x_batch, y_batch in data:
        train_step(model, x_batch, y_batch, lr)

def accuracy(y_pred, y_real):
    goods = np.sum(np.argmax(y_pred, axis=1) == np.argmax(y_real, axis=1))
    acc = goods / len(y_real)
    return acc
```

```python
X = X.reshape(X.shape[0], 28 * 28)

Y = np.zeros((train_labels.shape[0], 10))
for i, label in enumerate(train_labels):
    Y[i, label] = 1

batch_size = 500
num_samples = X.shape[0]

data = []
for i in range(0, num_samples, batch_size):
    x_batch = X[i:i + batch_size]
    y_batch = Y[i:i + batch_size]
    data.append((x_batch, y_batch))
#print(data)

model = Model([
    Deep(784, 100),           
    Activation(relu, relu_deriv),
    Deep(100, 10)              
])


lr = 0.03
epochs = 10
epoch = 0

while epoch < epochs:
    print("Epoch:", epoch + 1)
    train_loop(model, data, lr)
    epoch += 1
```
